## 2. Ficha de mi herramienta

### Nombre
`convertir_usd_a_cop`

### Descripción
Convierte un monto en **dólares estadounidenses (USD)** a **pesos colombianos (COP)** multiplicándolo por la TRM expresada en **COP por USD**.  
**Fuente:** monto y TRM ingresados manualmente por el usuario; no utiliza una API ni consulta internet.

### Parámetros

| Parámetro | Tipo | Formato | ¿Obligatorio? | Valor por defecto |
|---|---|---|---|---|
| `monto_usd` | `float` (número) | USD, ejemplo: `50000` | Sí | Ninguno |
| `trm` | `float` (número) | COP/USD, ejemplo: `3183.60` | Sí | Ninguno |
| `decimales` | `int` (entero) | Ejemplo: `2` | No | `2` |

### Estructura de la respuesta

La herramienta devuelve un **diccionario**, equivalente a un objeto JSON:

```json
{
  "estado": "ok",
  "monto_usd": 50000,
  "trm": 3183.60,
  "total_cop": 159180000.0,
  "moneda": "COP",
  "mensaje": "Conversion realizada"
}
```

**Fórmula:** `total_cop = monto_usd * trm`

### Si no hay datos o hay errores

| Situación | Qué devuelve |
|---|---|
| Falta el monto o la TRM | `estado: "sin_datos"`, `total_cop: null` y un mensaje |
| Monto, TRM o decimales incorrectos | `estado: "error_validacion"`, `total_cop: null` y un mensaje |
| La fuente no permite obtener un valor | Como los datos son manuales, se considera **sin datos** y no se inventa una TRM |

### ¿Por qué le sirve a otra persona?
Esta herramienta ayuda a estudiantes, comerciantes, contadores e importadores a estimar valores en pesos colombianos a partir de montos en dólares de manera rápida y organizada.

## 4. Crear mi herramienta en Python

Usaremos **`def`** para crear la herramienta, **`if`** para revisar los datos y **`return`** para devolver la respuesta. La operación principal sigue siendo una sola multiplicación.

- `isinstance()` comprueba que un dato sea numérico.
- `round()` redondea el resultado.
- `float`, `int` y `dict` describen los tipos de datos que utiliza la herramienta.

**No tienes que memorizar todo:** lo importante es identificar dónde se multiplican los dólares por la TRM.

````python

def convertir_usd_a_cop(monto_usd: float = None, trm: float = None, decimales: int = 2) -> dict:
    """Convierte USD a COP usando la TRM que escribe el usuario."""

    estado = "ok"
    mensaje = "Conversion realizada"
    total = None

    if monto_usd is None or trm is None:
        estado = "sin_datos"
        mensaje = "Falta monto o TRM"
    elif not isinstance(monto_usd, (int, float)) or not isinstance(trm, (int, float)) or not isinstance(decimales, int):
        estado = "error_validacion"
        mensaje = "Los datos deben ser números"
    elif monto_usd < 0 or trm <= 0 or decimales < 0:
        estado = "error_validacion"
        mensaje = "Revisa los valores ingresados"
    else:
        total = round(monto_usd * trm, decimales)

    return {
        "estado": estado,
        "monto_usd": monto_usd,
        "trm": trm,
        "total_cop": total,
        "moneda": "COP",
        "mensaje": mensaje
    }

print("Herramienta creada: convertir_usd_a_cop")
````

## 5. Usar la herramienta con un ejemplo

Hacemos el mismo cálculo que aprendimos: **50.000 USD × 3.183,60 COP/USD = 159.180.000 COP**.

````python
resultado = convertir_usd_a_cop(50000, 3183.60)

print("Resultado de la herramienta:")
print(resultado)
````

## 6. Presentar las conversiones en una tabla

Esta tabla permite comparar varios ejemplos. Cada fila se calcula llamando nuestra herramienta; **no escribimos los resultados a mano**.

Puedes cambiar los montos y las TRM para practicar.

````python
conversion_1 = convertir_usd_a_cop(50000, 3183.60)
conversion_2 = convertir_usd_a_cop(1000, 4000)
conversion_3 = convertir_usd_a_cop(250, 3900)

tabla = pd.DataFrame([conversion_1, conversion_2, conversion_3])

print("TABLA DE CONVERSIONES USD A COP")
display(tabla)
````

## 7. Comprobar qué pasa si hay un error

Una herramienta también debe **avisar cuando no puede calcular**. Probamos dos casos: falta la TRM y la TRM es negativa.

````python
sin_trm = convertir_usd_a_cop(50000)
trm_incorrecta = convertir_usd_a_cop(50000, -100)

tabla_errores = pd.DataFrame([sin_trm, trm_incorrecta])

print("TABLA DE VALIDACIÓN")
display(tabla_errores)
````
